# Regression

Loss function 

L1Loss(MAE) -> torch.nn.L1Loss 

L2Loss(MSE) -> torch.MSELoss

```python
#Ex1
import torch
from torch import nn
import matplotlib.pyplot as plt

device = "cuda" if torch.cuda.is_available() else "cpu"
print(device)
!nvidia-smi

X = torch.arange(1,101,1).unsqueeze(dim=1).to(device)
Weight = 0.3
bias = 0.9
y = Weight*X + bias

X_train = X[:int(len(X)*0.8)]
X_test = X[int(len(X)*0.8):]
y_train = y[:int(len(y)*0.8)]
y_test = y[int(len(y)*0.8):]

X_train = X_train.to(device)
X_test = X_test.to(device)
y_train = y_train.to(device)
y_test = y_test.to(device)

def PlotTrainTest(X_train,X_test,y_train,y_test):
    plt.figure()
    plt.scatter(X_train,y_train,color="red",label = "Train")
    plt.scatter(X_test,y_test,color="g",label = "Test")
    plt.xlabel("x")
    plt.ylabel("y")
    plt.title("Plot of y w.r.t. x")
    plt.legend()

PlotTrainTest(X_train.cpu(),X_test.cpu(),y_train.cpu(),y_test.cpu())
```

```python
#Ex2
torch.manual_seed(42)
class Model(nn.Module):
    def __init__(self):
        super().__init__()
        self.weight = nn.Parameter(torch.randn(1,requires_grad=True,dtype = torch.float))
        self.bias = nn.Parameter(torch.randn(1, requires_grad = True,dtype = torch.float))

    def forward(self,X):
        return self.weight*X + self.bias

MyModel = Model().to(device)
MyModel.state_dict()
```

```python
#Ex3
LF = torch.nn.L1Loss()
optimizer = torch.optim.SGD(MyModel.parameters(),lr=0.0001)
epochs = 500
ep=[]
los=[]
for epoch in range(epochs):
    MyModel.train()
    TrainPred = MyModel(X_train)
    Trainloss = LF(TrainPred, y_train)
    optimizer.zero_grad()
    Trainloss.backward()
    optimizer.step()
    MyModel.eval()
    with torch.inference_mode():
        TestPred = MyModel(X_test)
        Testloss = LF(TestPred,y_test)
        if (epoch%10==1):
            ep.append(epoch)
            los.append(Testloss)
```

```python
#plotting model prediction 
with torch.inference_mode():
    TestPred = MyModel(X_test)
    plt.figure()
    plt.scatter(X_test.cpu(),TestPred.cpu(),label="TestPred")
    plt.scatter(X_train.cpu(),y_train.cpu(),label = "Train")
    plt.scatter(X_test.cpu(),y_test.cpu(),label = "Test")
    plt.legend()
```

```python
print(MyModel.state_dict())
```
